# Gradient Descent
Forward pass:

$$\hat{y} = w \cdot x$$

Loss function:

$$L = (y - \hat{y})^2$$

Gradient `w`:

$$\frac{dL}{dw} = -2x(y - \hat{y})$$

Update rule:

$$w_{new} = w_{old} - \alpha \cdot \frac{dL}{dw}$$

```python
x = 2
y = 6

w = 0.0
learningRate = 0.1
epochs = 10
```

```python
def compute_loss(y, yHat):
    return (y - yHat) ** 2

def compute_gradient(x, y, yHat):
    return -2 * x * (y - yHat)

def forward_pass(x, w):
    return w * x

def update_weight(w, learningRate, gradient):
    return w - learningRate * gradient
```

```python
for epoch in range(1, epochs + 1):
    yHat = forward_pass(x, w)
    loss = compute_loss(y, yHat)
    gradient = compute_gradient(x, y, yHat)
    w = update_weight(w, learningRate, gradient)
    print(f"Epoch {epoch}: w = {w:.4f}, y_hat = {yHat:.4f}, loss = {loss:.4f}")

print(f"Final weight: {w:.4f}")
```
