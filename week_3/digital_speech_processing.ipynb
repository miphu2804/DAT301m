# Signal Generation and Plotting

```python
import numpy as np 
import matplotlib.pyplot as plt
import librosa 
```

## Load simulation parameters

```python
A = 1.5
omega = 0.1 * np.pi
phi = 0.3 * np.pi
N = 100 
```

```python
def generate_discrete_time_signals(A, omega, phi, N):
    n = np.arange(N)
    x_n = A * np.cos(omega * n + phi)
    return x_n, n
```

```python
x_n, n = generate_discrete_time_signals(A, omega, phi, N)
```

```python
plt.figure(figsize=(10, 6))
plt.stem(n, x_n)
plt.title('Discrete-Time Signal: $x[n] = A \cos(\omega n + \phi)$')
plt.xlabel('n (samples)')
plt.ylabel('x[n]')
plt.grid()
plt.show()
```

## Load sample audio file

```python
def load_audio_file(file_path):
    signal, sr = librosa.load(file_path, sr=None)
    return signal, sr
```
