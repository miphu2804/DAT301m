## Word based tokenization

### Word-based

```python
text = "I want to learn NLP using Python" 
```

```python
tokens = text.split()
print(tokens)
vocab = {token: id+1 for id, token in enumerate(tokens)}
print(vocab)
input_ids = [vocab[token] for token in tokens]
print(input_ids)
attention_masks = [1] * len(input_ids)
print(attention_masks)
```

### Subword-Based

```python
tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

sentences = [
    "dog runs fast",
    "dogs are running",
    "unicorns are magical"
]

for sentence in sentences:
    tokens = tokenizer.tokenize(sentence)
    ids = tokenizer.convert_tokens_to_ids(tokens)
    print(f"Sentence: {sentence}")
    print(f"Tokens: {tokens}")
    print(f"IDs: {ids}")
    print("---"*10)
```

## Character-Based

```python
sentences = [
    "dog runs fast",
    "dogs are running",
    "unicorns are magical"
]

for sentence in sentences:
    char_tokens = list(sentence.replace(" ", ""))
    print(f"Sentence: {sentence}")
    print(f"Char Tokens: {char_tokens}")
    print("---"*10)
```
